```python
import sys
import matplotlib
matplotlib.use('Agg')
```

```python

import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
x = [4, 5, 10, 4, 3, 11, 14 , 6, 10, 12]
y = [21, 19, 24, 17, 16, 25, 24, 22, 21, 21]
```

```python

data = list(zip(x, y))
```

```python

kmeans = KMeans(n_clusters=2)
```

```python
kmeans.fit(data)
```

```python

plt.scatter(x, y, c=kmeans.labels_)
```

```python
plt.show()
```
